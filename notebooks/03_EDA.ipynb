```python
# Convert date column

for col in df.columns:

    if "date" in col.lower():

        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

        date_column = col


print(date_column)
```

```python
# ==========================================
# Business KPIs
# ==========================================


kpi = {

"Total Records": len(df),

"Total Quantity": df["quantity"].sum()
if "quantity" in df.columns else None,

"Average Quantity":
df["quantity"].mean()
if "quantity" in df.columns else None

}


kpi
```

```python
# Create Revenue

if "price" in df.columns and "quantity" in df.columns:

    df["revenue"] = (
        df["price"] *
        df["quantity"]
    )


elif "sales_amount" in df.columns:

    df["revenue"] = df["sales_amount"]



df.head()
```

```python
total_revenue = df["revenue"].sum()

print(
    "Total Revenue:",
    total_revenue
)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["revenue"],
    bins=50
)

plt.title(
    "Revenue Distribution"
)

plt.xlabel(
    "Revenue"
)

plt.ylabel(
    "Frequency"
)

plt.show()
```

```python
if date_column:

    sales_time = (
        df
        .groupby(date_column)
        ["revenue"]
        .sum()
    )


    plt.figure(figsize=(12,5))

    sales_time.plot()

    plt.title(
        "Daily Revenue Trend"
    )

    plt.xlabel(
        "Date"
    )

    plt.ylabel(
        "Revenue"
    )

    plt.show()
```

```python
df["month"] = (
    df[date_column]
    .dt.month
)


monthly_sales = (
    df
    .groupby("month")
    ["revenue"]
    .sum()
)


plt.figure(figsize=(8,5))


monthly_sales.plot(
    kind="bar"
)


plt.title(
    "Monthly Revenue"
)


plt.show()
```

```python
if "sku_id" in df.columns:


    top_products = (
        df.groupby("sku_id")
        ["revenue"]
        .sum()
        .sort_values(
            ascending=False
        )
        .head(10)
    )


    top_products
```

```python
top_products.plot(
    kind="bar",
    figsize=(10,5)
)


plt.title(
    "Top 10 Products by Revenue"
)


plt.xlabel(
    "SKU"
)


plt.ylabel(
    "Revenue"
)


plt.show()
```
