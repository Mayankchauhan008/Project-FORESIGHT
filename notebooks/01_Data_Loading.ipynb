Load Library

```python
import os
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

 Data Path

```python
RAW_DATA_PATH = r"C:\Users\mayank\OneDrive\Desktop\Project-FORESIGHT\Data\raw"
PROCESSED_DATA_PATH = r"C:\Users\mayank\OneDrive\Desktop\Project-FORESIGHT\Data\cleaned"

print("Raw Data Path :", RAW_DATA_PATH)
print("Processed Path :", PROCESSED_DATA_PATH)
```

 Load CSV Files


```python
sales = pd.read_csv(RAW_DATA_PATH + "\\" + "bm_sales.csv")

inventory = pd.read_csv(RAW_DATA_PATH + "\\" + "bm_inventory.csv")

sku = pd.read_csv(RAW_DATA_PATH + "\\" + "bm_skus.csv")

stores = pd.read_csv(RAW_DATA_PATH + "\\" + "bm_stores.csv")

customers = pd.read_csv(RAW_DATA_PATH + "\\" + "bm_customers.csv")

promotions = pd.read_csv(RAW_DATA_PATH + "\\" + "bm_promotions.csv")
```

Store Dataframe in dictionary

```python
datasets = {
    "Sales": sales,
    "Inventory": inventory,
    "SKU": sku,
    "Stores": stores,
    "Customers": customers,
    "Promotions": promotions
}
```

Shape of data

```python
for name, df in datasets.items():
    print(f"{name}")
    print(df.shape)
```

See the record

```python
for name, df in datasets.items():

    print("\n")
    print("="*100)
    print(name)
    print("="*100)

    display(df.head())
```

Dataset information

```python
for name, df in datasets.items():

    print("="*60)
    print(name)
    print("="*60)

    print(df.info())
```

Check missing value

```python
missing_report = {}

for name, df in datasets.items():

    missing = df.isnull().sum()

    print("\n")

    print("="*50)

    print(name)

    print("="*50)

    print(missing)
```

Duplicate records

```python
for name, df in datasets.items():

    duplicates = df.duplicated().sum()

    print(f"{name} Duplicate Rows : {duplicates}")
```

Data type

```python
for name, df in datasets.items():

    print("="*50)

    print(name)

    print("="*50)

    print(df.dtypes)
```

Statistical Summary

```python
for name, df in datasets.items():

    print("="*60)

    print(name)

    print("="*60)

    display(df.describe(include='all'))
```

Column name

```python
for name, df in datasets.items():

    print("="*113)

    print(name)

    print("="*113)

    print(df.columns.tolist())
```

Memory use

```python
for name, df in datasets.items():

    memory = df.memory_usage(deep=True).sum() / 1024**2

    print(f"{name} Memory : {memory:.2f} MB")
```

Data Summary Table

```python
summary = []

for name, df in datasets.items():

    summary.append({

        "Dataset": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1],

        "Missing Values": df.isnull().sum().sum(),

        "Duplicates": df.duplicated().sum()

    })

summary_df = pd.DataFrame(summary)

summary_df
```

Save Summary

```python
summary_df.to_csv(

    PROCESSED_DATA_PATH + "\\"+"dataset_summary.csv",

    index=False
)

print("Dataset Summary Saved")
```

Primary Key Inspection

```python
for name, df in datasets.items():

    print("="*60)

    print(name)

    print("="*60)

    for col in df.columns:

        if df[col].is_unique:

            print(col, "-> Possible Primary Key")
```

Foreign Key Inspection

```python
common_columns = {}

for name, df in datasets.items():

    common_columns[name] = list(df.columns)

for name, cols in common_columns.items():

    print(name)

    print(cols)

    print()
```

Data Dictionary

```python
dictionary = []

for name, df in datasets.items():

    for col in df.columns:

        dictionary.append({

            "Dataset": name,

            "Column": col,

            "Data Type": df[col].dtype,

            "Missing": df[col].isnull().sum(),

            "Unique Values": df[col].nunique()

        })

dictionary_df = pd.DataFrame(dictionary)

dictionary_df.head()
```

Save Data Dictionary

```python
dictionary_df.to_csv(

    PROCESSED_DATA_PATH + "\\" + "data_dictionary.csv",

    index=False
)

print("Data Dictionary Saved")
```
